```python
#Asks the user for inputs
firstname = str(input("Enter first name: "))
lastname = str(input("Enter last name: "))
section = str(input("Enter section: "))
favoriteSubject = str(input("Enter favorite subject: "))

#The list
List = [firstname,lastname,section,favoriteSubject]

#The function
def printList():
  print("\nOutput:")

#A for loop that prints the outputs
  for i in range(len(List)):

    if i == 0:
      print("Name:",List[i],List[i+1])
      print("Name Code:",List[i],List[i+1])

    elif i == 1:
      print("Section:",List[i+1])

    elif i == 2:
      print("Favorite Subject:",List[i+1])

    elif i == 3:
      count = len(List[0] + List[1])
      print("Character count:",count)

    else:
      break

#Function call
printList()


```
